## Reading different types of CSV files using PySpark

```python
# Basic Read
df = spark.read.csv('/Volumes/my_files/my_schema/csv_files/my_files/sales.csv')

df.display()
```

```python
# Basic read using format
df = spark.read.format('csv').load('/Volumes/my_files/my_schema/csv_files/my_files/sales.csv')

df.display()
```

```python
# 3. Header including in no headers files
df = spark.read.format('csv').option('header', 'true').load('/Volumes/my_files/my_schema/csv_files/my_files/sales.csv')

df.display()
```

```python
# Header + Schema (Recognizing the data types automatically using infraschema = true)
df = spark.read\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .csv('/Volumes/my_files/my_schema/csv_files/my_files/sales.csv')

df.display()
```

## Providing schema


```python
# Simple custom schema
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING"

df = spark.read.schema(schema_str).csv('/Volumes/my_files/my_schema/csv_files/my_files/sales_no_header.csv')
df.display()
```

```python
# 2nd way little bit advance custom schema 
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType

schema = StructType([
    StructField('id', IntegerType(), True),
    StructField('customer_name', StringType(), True),
    StructField('my_product', StringType(), True),
    StructField('quantity', IntegerType(), True),
    StructField('price', DoubleType(), True),
    StructField('remarks', StringType(), True)
])

df = spark.read.schema(schema).csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_no_header.csv")

df.display()
```

## Different File Formats Handling
### delimiter, quotes, escape, multiline

```python
# delimiter
df = spark.read\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .option('delimiter', '|')\
    .csv('/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe.csv')

df.display()
```

```python
# quote
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter", "|")\
    .option("quote", "'")\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote.csv")

df.display()
```

```python
# escape
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter", "|")\
    .option("quote", "'")\
    .option("escape", "#")\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote_escape.csv")

df.display()
```

```python
# multiline
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter", "|")\
    .option("quote", "'")\
    .option("escape", "#")\
    .option("multiline", "true")\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote_escape_multiline.csv")

df.display()
```
