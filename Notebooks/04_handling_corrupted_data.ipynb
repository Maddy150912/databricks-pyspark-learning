## Handling corrupted data

### For json

```python
# - reading in default mode
df = spark.read.format("json").\
     load("/Volumes/my_files/my_schema/json_files/different_json_files/sales_corrupted.json")

df.display()
```

```python
# - dropmalformed - currepted data ignor drop that recored

df = spark.read.format("json")\
     .option("mode","DROPMALFORMED")\
     .load("/Volumes/my_files/my_schema/json_files/different_json_files/sales_corrupted.json")

df.display()
```

```python
# - Failfast
df = spark.read.format("json")\
    .option("mode", "FAILFAST")\
    .load("/Volumes/my_files/my_schema/json_files/different_json_files/sales_corrupted.json")

df.display()
```

```python
# - Permissive - default mode
df = spark.read.format("json")\
    .option("mode", "PERMISSIVE")\
    .load("/Volumes/my_files/my_schema/json_files/different_json_files/sales_corrupted.json")

df.display()
```

### For csv files 

```python
df = spark.read.format("csv").load("/Volumes/my_files/my_schema/csv_files/my_files/sales_corrupted.csv")

df.display()
```

```python
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING, _corrupt_record STRING"

df = spark.read.format("csv")\
    .schema(schema_str)\
    .load("/Volumes/my_files/my_schema/csv_files/my_files/sales_corrupted.csv")

df.display()
```

```python
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING"

df = spark.read.format("csv")\
    .schema(schema_str)\
    .option("mode","DROPMALFORMED")\
    .load("/Volumes/my_files/my_schema/csv_files/my_files/sales_corrupted.csv")

df.display()
```

```python
# - Option 2
schema_str = 'id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING'

df = spark.read\
    .schema(schema_str)\
    .format('csv')\
    .option('badRecordsPath','/Volumes/my_files/my_schema/csv_files/badRecordsPath')\
    .load("/Volumes/my_files/my_schema/csv_files/my_files/sales_corrupted.csv")

df.display() 
```
