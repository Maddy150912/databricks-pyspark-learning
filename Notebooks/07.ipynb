```python
from pyspark.sql.functions import expr
#df2 = df.select("name",expr("country as my_country"))
df2 = df.select("name",expr("concat(city,'_',country) as location"))
df2.display()
```

```python
from pyspark.sql.functions import expr
df2 = df.selectExpr("name", "country as my_country")
df2.display()

df3 = df.select("name",expr("concat(city, '_', country)").alias('location'))
df3.display()
```

```python
%sql
SELECT name, concat(city, '_', country) AS location  FROM samples.bakehouse.sales_franchises
```
