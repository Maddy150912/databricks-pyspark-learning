## Read and Write parquet files

```python
df = spark.read.json("/Volumes/my_files/my_schema/json_files/different_json_files/sales_single_line.json")

df.display()
```

```python
# - Writing parquet file

df.write.format("parquet").save("/Volumes/my_files/my_schema/parquet_files/parquet_file1")
```

```python
# - Reading parquet file
df2 = spark.read.format("parquet").load("/Volumes/my_files/my_schema/parquet_files/parquet_file1")

df2.display()
```
