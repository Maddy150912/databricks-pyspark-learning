# Basic DataFrame Operations in PySpark
### Covers: Reading table, display, limit, schema, writing data

```python
df = spark.read.table("samples.bakehouse.sales_franchises")
# Reading data from an existing table in Databricks
```

```python
df.show()
# show() → console view
```

```python
df.display()
# display() → Databricks UI (better visualization)
```

```python
df.write.saveAsTable("test.test_schema.test_table")
# Saving processed data as a table in Databricks
```

## Limit the number of records in DataFrame

```python
df2 = df.limit(10)
df2.count()
# Limiting dataset for quick analysis/testing
```

```python
df2.printSchema()
# Understanding data structure and column types
```
