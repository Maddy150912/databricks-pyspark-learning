## Handling Unstructured Data & File Metadata in PySpark

```python
# Reads file line by line
df = spark.read.text("/Volumes/my_files/my_schema/unstructured_files/")

df.display()
```

```python
# Reads entire file as single record
df = spark.read.text("/Volumes/my_files/my_schema/unstructured_files/",wholetext=True)

df.display()
```

```python
# Reads files like images, PDFs, etc. as binary
df = spark.read.format("binaryFile").load("/Volumes/my_files/my_schema/unstructured_files/")

df.display()
```

```python
# Extracting file-level metadata for tracking and auditing
df = spark.read.text("/Volumes/my_files/my_schema/unstructured_files/",wholetext=True)

df2 = df.selectExpr("value","_metadata.file_path", "_metadata.file_name", "_metadata.file_size", "_metadata.file_modification_time")

df2.display()
```

```python
df = spark.read.parquet("/Volumes/my_files/my_schema/parquet_files/parquet_file1")

df.display()
```

```python
# Metadata helps track source file even after processing
df = spark.read.parquet("/Volumes/my_files/my_schema/parquet_files/parquet_file1")

df2=df.selectExpr("id","price","product","quantity","remarks","_metadata.file_path","_metadata.file_name","_metadata.file_size","_metadata.file_modification_time")

df2.display()
```
