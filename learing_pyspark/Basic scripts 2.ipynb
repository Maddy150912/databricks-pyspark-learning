## Recognizing the data types also using the ".option('infraSchema' , 'true')"

```python
df = spark.read\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .csv('/Volumes/my_files/my_schema/csv_files/my_files/sales.csv')

df.display()
```

## Providing schema


```python
# simple way
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING"

df = spark.read.schema(schema_str).csv('/Volumes/my_files/my_schema/csv_files/my_files/sales_no_header.csv')
df.display()
```

```python
# second way
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType

schema = StructType([
    StructField('id', IntegerType(), True),
    StructField('customer_name', StringType(), True),
    StructField('my_product', StringType(), True),
    StructField('quantity', IntegerType(), True),
    StructField('price', DoubleType(), True),
    StructField('remarks', StringType(), True)
])

df = spark.read.schema(schema).csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_no_header.csv")

df.display()
```

## Files seprated by Pipe (|)

```python
df = spark.read\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .option('delimiter', '|')\
    .csv('/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe.csv')

df.display()
```

## Removing quotes  

```python
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter", "|")\
    .option("quote", "'")\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote.csv")

df.display()
```

## Reading files which also have escape charchter

```python
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter", "|")\
    .option("quote", "'")\
    .option("escape", "#")\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote_escape.csv")

df.display()
```

## Reading files which have some data spread in multiple lines

```python
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter", "|")\
    .option("quote", "'")\
    .option("escape", "#")\
    .option("multiline", "true")\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote_escape_multiline.csv")

df.display()
```
